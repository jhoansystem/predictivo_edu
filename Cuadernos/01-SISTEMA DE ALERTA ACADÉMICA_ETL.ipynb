## Paso 4: Limpieza de Datos.
Aplicamos las transformaciones necesarias para corregir inconsistencias.

### 4.1 Eliminar columnas innecesarias- `Unnamed: 25`: columna fantasma sin datos

```python
# 4.1 Eliminar columna unnamed
if 'Unnamed: 25' in df.columns:
    df = df.drop(columns=['Unnamed: 25'])
    print('Columna Unnamed: 25 eliminada.')
```

### 4.2 Corrección de intensidad horaria.
La columna `intensidad_semannal` contiene valores numéricos mezclados con nombres de materias. Convertimos a numérico.

```python
# 4.2 Limpiar intensidad horaria
print('Valores unicos antes:', df['intensidad_semannal'].unique()[:15])
df['intensidad_semannal'] = pd.to_numeric(df['intensidad_semannal'], errors='coerce')
df['intensidad_semannal'] = df['intensidad_semannal'].fillna(1).astype(int)
print('Valores unicos despues:', sorted(df['intensidad_semannal'].unique()))
```

### 4.3 Corrección de notas (> 5.0).

El sistema de evaluación colombiano usa escala 0.0 - 5.0. Algunas notas duplicadas aparecen con valores > 5.0.

```python
# 4.3 Corregir notas fuera de rango
sospechosas = (df['nota'] > 5.0).sum()
print(f'Notas > 5.0: {sospechosas}')
if sospechosas > 0:
    df.loc[df['nota'] > 5.0, 'nota'] = df.loc[df['nota'] > 5.0, 'nota'] / 2
    print(f'Notas corregidas. Nuevo maximo: {df["nota"].max()}')
```

### 4.4 Limpieza de refuerzo (valores datetime mezclados).

La columna `refuerzo` contiene valores `0,1,2,3,4` y fechas mal formateadas.

```python
# 4.4 Limpiar refuerzo
df['refuerzo'] = df['refuerzo'].astype(str).str.strip()
df.loc[df['refuerzo'].isin(['0','']), 'refuerzo'] = '0'
df.loc[~df['refuerzo'].isin(['0','1','2','3','4']), 'refuerzo'] = '0'
df['refuerzo'] = df['refuerzo'].astype(int)
print('Refuerzo values:', df['refuerzo'].value_counts().to_dict())
```

### 4.5 Limpieza de AreasQueNoPromuevenAnoCantidad.
Contiene valores 'S' (string) mezclados con enteros.

```python
# 4.5 Limpiar areas que no promueven
area_clean = df['AreasQueNoPromuevenAnoCantidad'].astype(str).str.strip()
area_clean = area_clean.replace({'S': '0', 's': '0'})
area_clean = pd.to_numeric(area_clean, errors='coerce').fillna(0)
df['AreasQueNoPromuevenAnoCantidad'] = area_clean.astype(int)
print('Distribucion:', df['AreasQueNoPromuevenAnoCantidad'].value_counts().to_dict())
```

### 4.6 Estandarización de columnas categóricas

```python
# 4.6 Estandarizar categoricas
# materia_promueve: solo S o N
df['materia_promueve'] = df['materia_promueve'].apply(
    lambda x: 'S' if str(x).strip() in ['S', 's'] else 'N'
)

# promovido: llenar nulos con N (asumimos no promovido)
df['promovido'] = df['promovido'].fillna('N')

# faltas: reemplazar nulos por 0
df['faltas'] = df['faltas'].fillna(0).astype(int)
df['FALTAS_CON_EXCUSA'] = df['FALTAS_CON_EXCUSA'].fillna(0).astype(int)

# notaoriginal: llenar con la nota actual si falta
df['notaoriginal'] = df['notaoriginal'].fillna(df['nota'])
df['valoracion_original'] = df['valoracion_original'].fillna(df['valoracion'])

print('Estandarizacion completada.')
```

## Paso 5: Clasificación de Asignaturas por Áreas del Saber.
Según la normativa colombiana, clasificamos cada asignatura en 10 áreas del saber.

```python
# 5. Clasificacion de asignaturas en areas del saber
# Criterios de clasificacion segun normativa colombiana:
# 1. CIENCIAS NATURALES Y EDUCACION AMBIENTAL
# 2. CIENCIAS SOCIALES, GEOGRAFIA, HISTORIA, ECONOMIA Y POLITICA
# 3. ARTES Y EDUCACION EN TECNOLOGIA, EMPRENDIMIENTO E INFORMATICA
# 4. EDUCACION FISICA, RECREACION Y DEPORTES
# 5. HUMANIDADES (LENGUA CASTELLANA, IDIOMA EXTRANJERO O INGLES)
# 6. MATEMATICAS
# 7. TECNICAS Y TECNOLOGICAS
# 8. OPTATIVAS Y CATEDRAS
# 9. ETICA, FILOSOFIA, EDUCACION RELIGIOSA
# 10. OTRAS

def clasificar_area(nommat):
    if pd.isna(nommat):
        return '10. OTRAS'
    nom = str(nommat).upper().strip().replace(' ', '')\
        .replace('Í','I').replace('Ó','O').replace('É','E')\
        .replace('Á','A').replace('Ú','U').replace('Ñ','N')

    if any(x in nom for x in ['CIENCIASNATURALES', 'QUIMICA', 'FISICA', 'BIOLOGIA', 'OPTATIVACIENCIASNATURALES']):
        return '1. CIENCIAS NATURALES Y EDUCACION AMBIENTAL'
    if any(x in nom for x in ['CIENCIASSOCIALES', 'HISTORIA', 'GEOGRAFIA', 'CONSTITUCION', 'CIENCIASPOLITICAS', 'ECONOMICAS']):
        return '2. CIENCIAS SOCIALES, GEOGRAFIA, HISTORIA, ECONOMIA Y POLITICA'
    if any(x in nom for x in ['ARTISTICA', 'TECNOLOGIA', 'INFORMATICA', 'EMPRENDIMIENTO']):
        return '3. ARTES Y EDUCACION EN TECNOLOGIA, EMPRENDIMIENTO E INFORMATICA'
    if any(x in nom for x in ['FISICA-RECREACION', 'EDUCACIONFISICA', 'DEPORTES']):
        return '4. EDUCACION FISICA, RECREACION Y DEPORTES'
    if any(x in nom for x in ['LENGUACASTELLANA', 'IDIOMAEXTRANJERO', 'INGLES', 'HUMANIDADES', 'PLANLECTOR', 'OPTATIVALENGUACASTELLANA', 'OPTATIVAINGLES']):
        return '5. HUMANIDADES (LENGUA CASTELLANA, IDIOMA EXTRANJERO O INGLES)'
    if any(x in nom for x in ['MATEMATICAS', 'GEOMETRIA', 'ESTADISTICA']):
        return '6. MATEMATICAS'
    if any(x in nom for x in ['TECNICO', 'LOGISTICA', 'SEGURIDADINDUSTRIAL', 'ASISTENCIAADMINISTRATIVA', 'RECURSOSHUMANOS', 'INVESTIGACION']):
        return '7. TECNICAS Y TECNOLOGICAS'
    if any(x in nom for x in ['OPTATIVA', 'CATEDRA']):
        return '8. OPTATIVAS Y CATEDRAS'
    if any(x in nom for x in ['ETICA', 'VALORESHUMANOS', 'FILOSOFIA', 'RELIGIOSA', 'PROYECTODEVIDA']):
        return '9. ETICA, FILOSOFIA, EDUCACION RELIGIOSA'
    return '10. OTRAS'

df['area_saber'] = df['nommat'].apply(clasificar_area)
print('Distribucion por areas:')
print(df['area_saber'].value_counts())
```

## Paso 6: Feature Engineering (Variables Agregadas).
Transformamos datos a nivel **estudiante-año** (cada fila = un estudiante en un año lectivo).

```python
# 6.1 Variables de rendimiento academico
# Promedio de notas por estudiante-año
promedio_anual = df.groupby(['matricula', 'ano'], as_index=False)['nota'].mean()
promedio_anual.columns = ['matricula', 'ano', 'promedio_nota']

# Nota minima y maxima
nota_min = df.groupby(['matricula', 'ano'], as_index=False)['nota'].min()
nota_min.columns = ['matricula', 'ano', 'nota_minima']
nota_max = df.groupby(['matricula', 'ano'], as_index=False)['nota'].max()
nota_max.columns = ['matricula', 'ano', 'nota_maxima']

# Desviacion estandar de notas (consistencia)
nota_std = df.groupby(['matricula', 'ano'], as_index=False)['nota'].std()
nota_std.columns = ['matricula', 'ano', 'nota_std']

# Materias perdidas (nota < 3.0 = desempeno bajo)
df['materia_perdida'] = (df['nota'] < 3.0).astype(int)
materias_perdidas = df.groupby(['matricula', 'ano'], as_index=False)['materia_perdida'].sum()
materias_perdidas.columns = ['matricula', 'ano', 'materias_perdidas']

print('Variables de rendimiento creadas.')
```

```python
# 6.2 Variables de asistencia
# Intensidad horaria total
intensidad_total = df.groupby(['matricula', 'ano'], as_index=False)['intensidad_semannal'].sum()
intensidad_total.columns = ['matricula', 'ano', 'intensidad_total_semanal']

# Total de faltas
faltas_total = df.groupby(['matricula', 'ano'], as_index=False)['faltas'].sum()
faltas_total.columns = ['matricula', 'ano', 'total_faltas']

# Porcentaje de inasistencia
merged_f = pd.merge(intensidad_total, faltas_total, on=['matricula', 'ano'])
merged_f['porc_inasistencia'] = (merged_f['total_faltas'] /
                                   merged_f['intensidad_total_semanal'].replace(0, 1)) * 100

print(f'Inasistencia media: {merged_f["porc_inasistencia"].mean():.2f}%')
print(f'Estudiantes con >33% inasistencia: {(merged_f["porc_inasistencia"] > 33).sum()}')
```

```python
# 6.3 Variables academicas adicionales
# Cantidad de asignaturas cursadas
asignaturas_count = df.groupby(['matricula', 'ano'], as_index=False)['codmat'].nunique()
asignaturas_count.columns = ['matricula', 'ano', 'cantidad_asignaturas']

# Total de refuerzos academicos
df['tuvo_refuerzo'] = (df['refuerzo'] > 0).astype(int)
refuerzos_count = df.groupby(['matricula', 'ano'], as_index=False)['tuvo_refuerzo'].sum()
refuerzos_count.columns = ['matricula', 'ano', 'total_refuerzos']

# Promedio por area del saber
promedio_area = df.groupby(['matricula', 'ano', 'area_saber'], as_index=False)['nota'].mean()
promedio_area_pivot = promedio_area.pivot_table(
    index=['matricula', 'ano'],
    columns='area_saber',
    values='nota',
    aggfunc='mean'
).reset_index()
# Renombrar columnas
area_cols = {col: f'prom_{col[:3].replace(" ","")}' if col not in ['matricula','ano'] else col
             for col in promedio_area_pivot.columns}
promedio_area_pivot = promedio_area_pivot.rename(columns=area_cols)

print('Variables academicas creadas.')
```

```python
# 6.4 Variables target y demograficas
# Grado y grupo
datos_estudiante = df[['matricula', 'ano', 'grado', 'grupo']].drop_duplicates(subset=['matricula', 'ano'])

# Target: promovido (S/N) y version binaria
promovido_target = df[['matricula', 'ano', 'promovido']].drop_duplicates(subset=['matricula', 'ano'])
promovido_target['promovido_bin'] = (promovido_target['promovido'] == 'S').astype(int)

print(f'Total estudiantes-año: {len(datos_estudiante)}')
print(f'Distribucion target:\n{promovido_target["promovido"].value_counts()}')
```

### 6.5 Variables de Coherencia y Contexto (Grupos y Cohortes).

Añadimos una variable de 'cohorte' que combina el año y el grado, lo que permite identificar grupos de estudiantes que avanzan juntos a través del sistema educativo. La variable `grupo` ya está incluida en los datasets agregados.

```python
import pandas as pd

# Ensure 'df' is available, assuming it has been loaded previously in the notebook
# If 'df' is not defined, this cell will raise an error. Ideally, the entire notebook should be run sequentially.

# Re-create datos_estudiante to ensure it's available for this cell
datos_estudiante = df[['matricula', 'ano', 'grado', 'grupo']].drop_duplicates(subset=['matricula', 'ano'])

# Crear una variable de cohorte (combinacion de año y grado)
# Esto identifica a los estudiantes que estan en el mismo grado en el mismo año academico
datos_estudiante['cohorte'] = datos_estudiante['ano'].astype(str) + '-' + datos_estudiante['grado'].astype(str)

print('Variable "cohorte" creada en datos_estudiante.')
display(datos_estudiante.head())
```

## Paso 7: Construcción de Datasets para Cada Modelo.
Creamos **3 datasets** específicos, cada uno optimizado para un tipo de modelo:
1. **Regresión Lineal**: Target = `promedio_nota` (numérico continuo)
2. **Regresión Logística**: Target = `promovido_bin` (binario 0/1)
3. **Árboles de Decisión**: Target = `promovido_bin` + características de interacción

```python
# 7.1 Dataset para Regresion Lineal
# Target: promedio_nota (predecir nota promedio del estudiante)
df_lr = datos_estudiante.copy()

# Funcion de merge helper
def merge_features(base, other, keys=['matricula', 'ano']):
    cols = [c for c in other.columns if c in keys or c not in base.columns]
    return base.merge(other[cols], on=keys, how='left')

for mdf in [promedio_anual, intensidad_total, asignaturas_count, faltas_total,
            merged_f[['matricula','ano','porc_inasistencia']],
            materias_perdidas, refuerzos_count, nota_min, nota_max, nota_std]:
    df_lr = merge_features(df_lr, mdf)
df_lr = merge_features(df_lr, promedio_area_pivot)

df_lr.to_csv('dataset_regresion_lineal.csv', index=False)
print(f'Dataset RL: {df_lr.shape}')
print(f'Target: promedio_nota (rango: {df_lr["promedio_nota"].min():.2f} - {df_lr["promedio_nota"].max():.2f})')
df_lr.head()
```

```python
# 7.2 Dataset para Regresion Logistica
# Target: promovido_bin (0=No promovido, 1=Promovido)
df_log = datos_estudiante.copy()

for mdf in [promedio_anual, intensidad_total, asignaturas_count, faltas_total,
            merged_f[['matricula','ano','porc_inasistencia']],
            materias_perdidas, refuerzos_count, nota_min, nota_max, nota_std]:
    df_log = merge_features(df_log, mdf)
df_log = merge_features(df_log, promedio_area_pivot)
df_log = merge_features(df_log, promovido_target[['matricula','ano','promovido','promovido_bin']])

df_log.to_csv('dataset_regresion_logistica.csv', index=False)
print(f'Dataset RLog: {df_log.shape}')
print(f'Distribucion target:')
print(df_log['promovido_bin'].value_counts())
df_log.head()
```

```python
# 7.3 Dataset para Arboles de Decision
# Mismas variables que RL + caracteristicas de interaccion
df_dt = df_log.copy()

# Caracteristicas de interaccion (arboles las aprovechan mejor)
df_dt['promedio_x_asignaturas'] = df_dt['promedio_nota'] * df_dt['cantidad_asignaturas']
df_dt['faltas_x_materias_perdidas'] = df_dt['total_faltas'] * df_dt['materias_perdidas']

df_dt.to_csv('dataset_arboles_decision.csv', index=False)
print(f'Dataset AD: {df_dt.shape}')
df_dt.head()
```

## Paso 8: Resumen del Proceso ETL

### Descripción del Proceso ETL

El proceso se ha estructurado en tres fases fundamentales para garantizar la calidad de los datos destinados al modelado predictivo:

1.  **Extracción (Extraction):** Se consolidó la información histórica de 10 años (2015-2024) proveniente de sistemas de registro académico. Se cargaron más de 330,000 registros de calificaciones individuales.
2.  **Transformación (Transformation):**
    *   **Limpieza:** Se corrigieron intensidades horarias, notas fuera de rango (escala 0-5) y se estandarizaron variables categóricas.
    *   **Clasificación:** Se categorizaron más de 50 asignaturas en las 10 áreas del saber según la normativa colombiana.
    *   **Ingeniería de Características:** Se realizó una agregación a nivel estudiante-año, calculando promedios, porcentajes de inasistencia, conteo de áreas reprobadas y creación de cohortes.
3.  **Carga (Loading):** Se generaron archivos CSV especializados y un dataset maestro limpio, listos para ser consumidos por algoritmos de Regresión Lineal, Logística y Árboles de Decisión.

### Criterios de NO PROMOCIÓN (Sistema Educativo Colombiano)

| Criterio | Descripción | Valor límite |
| :--- | :--- | :--- |
| **Nota mínima** | Desempeño bajo | < 3.0 (escala 0.0-5.0) |
| **Áreas perdidas** | Mínimo de áreas o asignaturas reprobadas | ≥ 3 áreas con nota < 3.0 |
| **Inasistencia** | Supera el límite permitido de fallas | > 33% de faltas sobre intensidad total |
| **Refuerzos** | Actividades de recuperación pedagógica | Aplica cuando la nota es < 3.0 |

### Variables Generadas para Modelado

| Variable | Tipo | Descripción |
| :--- | :--- | :--- |
| `promedio_nota` | Continua | Promedio anual del estudiante |
| `intensidad_total_semanal` | Discreta | Horas semanales totales cursadas |
| `cantidad_asignaturas` | Discreta | Número total de materias inscritas |
| `total_faltas` | Discreta | Sumatoria de inasistencias en el año |
| `porc_inasistencia` | Continua | Porcentaje de inasistencia relativo a la carga horaria |
| `materias_perdidas` | Discreta | Conteo de materias con nota < 3.0 |
| `total_refuerzos` | Discreta | Cantidad de procesos de refuerzo realizados |
| `nota_minima` / `maxima` | Continua | Notas extremas alcanzadas por el alumno |
| `nota_std` | Continua | Desviación estándar de notas (mide consistencia) |
| `prom_*` | Continua | Promedios específicos por cada una de las áreas del saber |
| `cohorte` | Categórica | Identificador de grupo por Año y Grado |

### Datasets Generados
1. `dataset_regresion_lineal.csv`: Optimizado para predecir el promedio numérico.
2. `dataset_regresion_logistica.csv`: Optimizado para clasificación binaria (Promovido S/N).
3. `dataset_arboles_decision.csv`: Incluye variables de interacción para modelos no lineales.
4. `Resultados_2015-2024_limpio.csv`: Registro histórico completo depurado.
